```python
def clean_text(text):
    text = text.replace("\r\n", "\n")  # 윈도우 줄바꿈 → 일반 줄바꿈
    text = re.sub(r"[ \t]+\n", "\n", text)  # 줄 끝 공백 지우기
    text = re.sub(r"\n{3,}", "\n\n", text)  # 빈 줄이 3개 이상이면 1개로
    return text.strip()  # 맨 앞뒤 공백 지우기

sample = "## 제40조 대피명령   \n\n\n\n- 대피명령을 받은 경우에는 즉시 명령에 따라야 한다.   \n"
print(repr(sample))
print(repr(clean_text(sample)))
```

```python
from collections import defaultdict

DOCUMENTS = Path("documents")  # rag 폴더 안에 있는 documents 폴더

def normalize(text):
    return re.sub(r"\s+", " ", text).strip()  # 모든 공백·줄바꿈을 띄어쓰기 하나로

seen = {}  # 정규화한 본문 → 처음 나온 위치
dups = defaultdict(list)  # 정규화한 본문 → 중복된 위치들

for p in sorted(DOCUMENTS.rglob("*.md")):  # 하위 폴더까지 .md 파일만 모음
    text = p.read_text(encoding="utf-8")
    for part in re.split(r"\n(?=## )", text):  # ## 단위로 청크 나누기
        if not part.startswith("## "):
            continue
        section, _, body = part.partition("\n")  # 제목과 본문 나누기
        key = normalize(body)  # 공백 무시하고 비교하기
        loc = f"{p.relative_to(DOCUMENTS)} / {section}"
        if key in seen:
            dups[key].append(loc)  # 이미 본 본문이면 중복으로 기록
        else:
            seen[key] = loc

for key, locs in dups.items():
    print("중복:", seen[key], "→", locs)
print("중복 그룹 수:", len(dups))
```

```python
chunks = []  # 청크들을 담을 빈 리스트
for d in docs:  # 문서 14개를 하나씩
    for part in re.split(r"\n(?=## )", d["text"]):  # ## 단위로 나누기
        if not part.startswith("## "):
            continue
        section, _, body = part.partition("\n")  # 제목과 본문 나누기
        chunks.append({"file": d["file"], "section": section.lstrip("# ").strip(), "text": body.strip()})

print(f"청크: {len(chunks)}개")
```

```python
docs = []
for p in sorted(DOCUMENTS.rglob("*.md")):
    text = clean_text(p.read_text(encoding="utf-8"))
    head = text.split("\n## ")[0]  # 첫 번째 ## 앞 머리말만
    urls = re.findall(r"https?://[^\s\)\]]+", head)  # 머리말에서 링크 뽑기
    m_출처 = re.search(r"출처[:：]\s*(.+)", head)
    m_시행 = re.search(r"시행\s*([0-9.]+)", head)
    m_확인 = re.search(r"확인일[:：]\s*([0-9\-]+)", head)
    docs.append({
        "file": str(p.relative_to(DOCUMENTS)),
        "title": text.splitlines()[0].lstrip("# "),
        "text": text,
        "형식": "md",
        "출처": m_출처.group(1).strip() if m_출처 else None,
        "시행일": m_시행.group(1) if m_시행 else None,
        "확인일": m_확인.group(1) if m_확인 else None,
        "source_url": urls[0] if urls else None,
    })
```
